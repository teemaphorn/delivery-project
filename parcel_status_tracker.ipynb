```python
import random
import pandas as pd

df_intake = pd.read_csv("parcel_intake_300.csv")

def generate_customer_name():
    first_names = ["เรยา", "มุนินทร์", "มุตตา", "เรณู", "พิไล", "ทองดี"]
    last_names = ["วงศ์เสวต", "จงสวัสดิ์", "ตาคลี", "อัศวรุ่งเรืองกิจ", "บุญพูลทรัพย์"]
    return f"{random.choice(first_names)} {random.choice(last_names)}"

class ParcelStatusTracker:

    def __init__(self, intake_id, parcel_id, recipient_name, old_status):
        self.intake_id = intake_id
        self.parcel_id = parcel_id
        self.recipient_name = recipient_name
        self.old_status = old_status
        self.new_status = old_status

    def update_status(self, target_status):
        valid_statuses = ["รับแล้ว", "กำลังส่ง", "ส่งสำเร็จ"]
        if target_status in valid_statuses:
            self.new_status = target_status

            print(
                f" [UPDATE] พัสดุ {self.parcel_id} ({self.recipient_name} ) :"
                f" '{self.old_status}' ➔ '{self.new_status}'"
            )


status_options = ["รับแล้ว", "กำลังส่ง", "ส่งสำเร็จ"]
old_statuses = []
new_statuses = []
new_recipient_names = []  

print(" อัปเดตสถานะพัสดุทั้งหมด 300 รายการ ")

for idx, row in df_intake.iterrows():

    thai_name = generate_customer_name()

    tracker = ParcelStatusTracker(
        intake_id=row["intake_id"],
        parcel_id=row["parcel_id"],
        recipient_name=thai_name, 
        old_status="รับแล้ว",
    )

    next_status = random.choice(status_options)
    tracker.update_status(next_status)

    new_recipient_names.append(thai_name)
    old_statuses.append(tracker.old_status)
    new_statuses.append(tracker.new_status)


df_intake["recipient_name"] = new_recipient_names
df_intake["old_status"] = old_statuses
df_intake["updated_status"] = new_statuses


columns_order = [
    "intake_id",
    "parcel_id",
    "sender_name",
    "recipient_name",
    "weight_kg",
    "service_type",
    "distance_km",
    "old_status",
    "updated_status",
]
```
