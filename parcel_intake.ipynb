```python
#(ส่วนรับเข้าพัสดุทีละชิ้น)
import pandas as pd

class Parcel:
    def __init__(self, parcel_id, sender, weight_kg, distance_km, status="Received"):
        self.parcel_id = parcel_id
        self.sender = sender
        self.weight_kg = float(weight_kg)
        self.distance_km = float(distance_km)
        self.status = status
        self.shipping_cost = self.calculate_cost()

    def calculate_cost(self, base_fare=30.0):
        """คำนวณค่าจัดส่งตามน้ำหนักและระยะทาง"""
        cost = base_fare + (self.weight_kg * 10.0) + (self.distance_km * 5.0)
        return round(cost, 2)


class ParcelIntakeManager:
    def __init__(self):
        self.intake_list = []

    def receive_single_parcel(self, parcel_id, sender, weight_kg, distance_km):
        """ฟังก์ชันสำหรับรับเข้าพัสดุทีละ 1 ชิ้น"""
        new_parcel = Parcel(parcel_id, sender, weight_kg, distance_km)

        # แปลงเป็น Dictionary เพื่อเตรียมส่งต่อ
        parcel_data = {
            "parcel_id": new_parcel.parcel_id,
            "sender": new_parcel.sender,
            "weight_kg": new_parcel.weight_kg,
            "distance_km": new_parcel.distance_km,
            "shipping_cost": new_parcel.shipping_cost,
            "status": new_parcel.status
        }

        self.intake_list.append(parcel_data)
        print(f"✅ บันทึกพัสดุสำเร็จ: {parcel_id} | ผู้ส่ง: {sender} | ค่าส่ง: {new_parcel.shipping_cost} THB")
        return parcel_data

    def get_dataframe(self):
        """แปลงรายการพัสดุที่รับเข้าทั้งหมดเป็น DataFrame ให้เพื่อนนำไปทำ SQL/Pandas ต่อ"""
        return pd.DataFrame(self.intake_list)


# --- ตัวอย่างการรันรับเข้าพัสดุทีละชิ้น ---
manager = ParcelIntakeManager()

# รับเข้าพัสดุชิ้นที่ 1
manager.receive_single_parcel(parcel_id="P001", sender="Somchai", weight_kg=2.5, distance_km=12.0)

# รับเข้าพัสดุชิ้นที่ 2
manager.receive_single_parcel(parcel_id="P002", sender="Apinya", weight_kg=0.8, distance_km=5.5)

# แสดงผลลัพธ์พัสดุที่รับเข้าแล้ว
df_intake_result = manager.get_dataframe()
display(df_intake_result)
```
