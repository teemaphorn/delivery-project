```python
# ส่วนที่ 1:ระบบรับเข้าพัสดุทีละชิ้น
class Parcel:
    def __init__(self, parcel_id, recipient, weight, distance):
        self.parcel_id = parcel_id
        self.recipient = recipient
        self.weight = weight
        self.distance = distance
        self.status = "รับแล้ว"  
        
parcels_db = {}

def add_single_parcel(parcel_id, recipient, weight, distance):
    if parcel_id in parcels_db:
        print(f"ข้อผิดพลาด: เลขพัสดุ {parcel_id} มีในระบบแล้ว")
        return
        
    parcel_obj = Parcel(parcel_id, recipient, weight, distance)
    parcels_db[parcel_id] = parcel_obj
    print(f"บันทึกพัสดุ {parcel_id} สำเร็จ! (สถานะ: {parcel_obj.status})")

def show_all_parcels():
    print("\n--- รายการพัสดุในระบบ ---")
    if not parcels_db:
        print("ยังไม่มีข้อมูล")
        return
        
    for pid, obj in parcels_db.items():
        print(f"เลขพัสดุ: {pid} | ผู้รับ: {obj.recipient} | น้ำหนัก: {obj.weight} กก. | ระยะทาง: {obj.distance} กม.")

# --- ตัวอย่างการทดสอบเรียกใช้งานทีละรายการ ---
add_single_parcel("TH101", "มุนิน", 2.0, 120.0)
add_single_parcel("TH102", "มุตา", 3.5, 90.0)
show_all_parcels()
```
